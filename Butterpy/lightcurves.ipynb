# Generating A Lightcurve Using ``Butterpy``

```python
# importing butterpy packages
import butterpy as bp
import matplotlib.pyplot as plt
import numpy as np
%matplotlib inline
```

```python
# emergence active regions -- butterfly like patterb
np.random.seed(88)

star = bp.Surface()
star.emerge_regions(
    butterfly=True, 
    activity_level=3, 
    cycle_period=11,
    cycle_overlap=2, 
    max_lat=50, 
    min_lat=10,
)

time = np.arange(40, 200, 1.0)

# spot evolution and lightcurve
lightcurve = star.evolve_spots(
  time,
  alpha_med=1e-4, 
  period=24.5,
  inclination=60, 
  tau_evol=10,
  shear=0.25,
)
```

```python
# saving lc to a csv file
from astropy.table import Table

def lc_to_csv(lc, filename):
    table = Table(data = [lc.time, lc.flux], names = ['time', 'flux'])
    table.write(filename, format='csv', overwrite=True)

lc = lightcurve
lc_to_csv(lc, 'lc.csv')
```
